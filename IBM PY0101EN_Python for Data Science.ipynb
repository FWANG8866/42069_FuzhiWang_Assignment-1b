```python

import pandas as pd

# 1. Approximate RGM experimental data
# Digitised from published figures
data = {
    'Rubber_Content (%)': [0, 10, 20, 30, 40] * 3,
    'Confining_Pressure (kPa)': [10]*5 + [40]*5 + [70]*5,
    'Peak_Deviator_Stress (kPa)': [
        155, 135, 115, 82, 75,
        350, 295, 255, 250, 220,
        525, 410, 355, 350, 315
    ]
}

# 2. Create a Pandas DataFrame
df = pd.DataFrame(data)

# 3. Save and read CSV file
df.to_csv('RGM_data.csv', index=False)
df = pd.read_csv('RGM_data.csv')

# 4. Display the dataset
print("Original dataset:")
print(df.head())

# 5. Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# 6. Filter data by rubber content
filtered = df[df['Rubber_Content (%)'] == 20]

print("\nFiltered dataset (20% rubber):")
print(filtered)

# 7. Calculate stress reduction
df['Stress_Reduction (%)'] = df.groupby(
    'Confining_Pressure (kPa)'
)['Peak_Deviator_Stress (kPa)'].transform(
    lambda x: (x.iloc[0] - x) / x.iloc[0] * 100
).round(2)

# 8. Summarise peak deviator stress
summary = df.pivot(
    index='Rubber_Content (%)',
    columns='Confining_Pressure (kPa)',
    values='Peak_Deviator_Stress (kPa)'
)

print("\nPeak deviator stress summary:")
print(summary)

print("\nStress reduction results:")
print(df)

# 9. Export processed results
df.to_csv('RGM_processed_data.csv', index=False)
summary.to_csv('RGM_summary.csv')

print("\nData processing completed successfully.")
```
