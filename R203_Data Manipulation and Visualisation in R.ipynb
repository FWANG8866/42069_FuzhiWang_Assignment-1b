```python

# ============================================
# RGM data manipulation and visualisation
# R203 practical exercise
# ============================================

# 1. Install and load required packages
packages <- c("dplyr", "tidyr", "ggplot2")

for (pkg in packages) {
  if (!requireNamespace(pkg, quietly = TRUE)) {
    install.packages(
      pkg,
      repos = "https://cloud.r-project.org"
    )
  }
}

library(dplyr)
library(tidyr)
library(ggplot2)

# 2. Create approximate RGM experimental dataset
# Values digitised from published figures
# Peak deviator stress in kPa

rgm_data <- data.frame(
  Rubber_Content = c(0, 10, 20, 30, 40),
  P10 = c(155, 135, 115, 82, 75),
  P40 = c(350, 295, 255, 250, 220),
  P70 = c(525, 410, 355, 350, 315)
)

print("Original dataset:")
print(rgm_data)

# 3. Reshape data using tidyr
data_long <- rgm_data %>%
  pivot_longer(
    cols = c(P10, P40, P70),
    names_to = "Confining_Pressure",
    values_to = "Peak_Stress"
  ) %>%
  mutate(
    Confining_Pressure = factor(
      Confining_Pressure,
      levels = c("P10", "P40", "P70"),
      labels = c("10 kPa", "40 kPa", "70 kPa")
    )
  )

# 4. Check missing values
print("Missing values:")
print(colSums(is.na(data_long)))

data_long <- drop_na(data_long)

# 5. Process data using dplyr
data_summary <- data_long %>%
  group_by(Confining_Pressure) %>%
  arrange(Rubber_Content, .by_group = TRUE) %>%
  mutate(
    Stress_Reduction = round(
      (first(Peak_Stress) - Peak_Stress) /
        first(Peak_Stress) * 100,
      2
    )
  ) %>%
  ungroup()

# 6. Display processed data
print("Processed RGM data:")
print(data_summary)

# 7. Create visualisation using ggplot2
p <- ggplot(
  data_summary,
  aes(
    x = Rubber_Content,
    y = Peak_Stress,
    color = Confining_Pressure,
    group = Confining_Pressure
  )
) +
  geom_line(linewidth = 1) +
  geom_point(size = 3) +
  scale_x_continuous(
    breaks = c(0, 10, 20, 30, 40)
  ) +
  labs(
    x = "Rubber Content (%)",
    y = "Peak Deviator Stress (kPa)",
    color = "Confining Pressure",
    title = "Effect of Rubber Content on Peak Deviator Stress"
  ) +
  theme_classic(base_size = 13) +
  theme(
    plot.title = element_text(hjust = 0.5),
    legend.position = "right"
  )

# 8. Display figure
print(p)

# 9. Export processed data
write.csv(
  data_summary,
  "RGM_Experimental_Data_R.csv",
  row.names = FALSE
)

# 10. Save figure
ggplot2::ggsave(
  filename = "RGM_Analysis_R.png",
  plot = p,
  width = 8,
  height = 6,
  dpi = 300
)

# 11. Confirm saved files
print(getwd())
print(file.exists("RGM_Experimental_Data_R.csv"))
print(file.exists("RGM_Analysis_R.png"))

cat("RGM data analysis completed successfully.\n")
```
